### Peki, bir yapay sinir ağı, kendisine verilen kelime dizisini nasıl takip ederek bir sonraki kelimeyi tahmin edebilir?

Ağın içindeki milyarlarca parametre bulunuyor ve bu parametreleri nasıl optimize edeceğimizi anlamış durumdayız. Ancak, bu milyarlarca parametrenin tam olarak ne iş yaptığı konusununda bir belirsizlik söz konusu. Bir sonraki tahmini nasıl daha iyi yapacağımızı biliyoruz ancak parametrelerin tahmin sürecinde nasıl birbirleriyle etkileşim içinde olduğunu anlama konusunda eksikliklerimiz var

Bu parametreler, sinir ağının içinde bir tür bilgi veritabanı işlevi görür, fakat bu veritabanı kusursuz değildir ve bize bazen tuhaf sonuçlar sunabilir. Örneğin, ChatGPT ye 'Tom Cruise'un annes kim?' diye sorduğumuzda bize doğru bir şekilde 'Mary Lee Pfeiffer' yanıtını verecektir. Ancak 'Mary Lee Pfeiffer'ın oğlu kim?' diye sorduğumuzda 'bilmiyorum' yanıtını alabiliriz. Bu duruma 'reversal curse' ismi verilir ve gösteriyor ki ağın bilgisi bazen tek yönlü olabilir ve istikrarsız sonuçlara yol açabilir.

Sonuç olarak, LLM'leri anlaşılamayan, insan yapımı eserler olarak tanımlayabiliriz. Geleneksel mühendislik ürünleri gibi değiller; yani nasıl çalıştığını her yönüyle anladığımız bir otomobil gibi değil. Bunlar, çok sayıda optimizasyon aşamasından süzülerek ortaya çıkmış sinir ağları ve işleyiş mekanizmalarını tam olarak kavrayabilmiş değiliz

#### LLM nasıl eğitilir?

İndir, Eğit, Özel Instructions yaz

İndir: İlk olarak, geniş çapta dil verilerini içeren veri setlerini toplayın ve modelinizin eğitimi için hazır bir hale getirin.

Eğit: İkinci adımda, modelinizi bu veri setleri üzerinde eğiterek dilin genel yapısını, gramerini ve kullanımını öğrenmesini sağlayın.

Özel Talimatlar Yaz: Son olarak, modelinizi bir alana özelleştirmek için özel talimatlar ve rehberler yazın. Bu şekilde model, belirli görevleri yerine getirirken daha doğru ve etkili olacaktır.

##### Pretraining:

Bu aşama, genel dil anlama yeteneği kazandırmak için geniş bir veri kümesi üzerinde gerçekleştirilir. Model, genellikle büyük metin kümeleri ile önceden eğitilir. Bu süreçte, model dil yapısını, kelime ilişkilerini, cümle yapılarını anlama gibi genel dil özelliklerini öğrenir.

##### Fine-tuning:

Bu aşama ise önsel eğitimden elde edilen genel bilgiyi, belirli bir görev veya alan için daha spesifik hale getirmek üzere yapılır. Örneğin, bu genel dil modeli daha sonra özel bir alan için (örneğin, tıp, hukuk veya finans gibi) daha iyi performans gösterebilmesi için ilgili alana özgü veri setleriyle veya görevleriyle eğitilir. Bu sayede, genel dil modeli özel alandaki terminoloji, bağlamlar veya gereksinimlerle daha uyumlu hale gelir ve belirli bir alandaki görevleri daha iyi gerçekleştirebilir hale gelir.
