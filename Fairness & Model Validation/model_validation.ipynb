Model validation, modelimizin gerçek dünya verilerinde nasıl performans göstereceğini anlamak ve overfitting problemini tespit etmektir.

### **T-Test**

Ne zaman kullanılır: İki grup arasında fark olup olmadığını anlamak istediğimizde

Örnek: Yeni bir ilaç deniyorsunuz. İlaç alan ve almayan hastaların iyileşme sürelerini karşılaştırmak istiyorsunuz

Yorumlama: 0.05 ten küçükse gerçekten bir fark var diyebiliriz

### **Binomial Test**

Ne zaman kullanılır: Evet/Hayır, Başarılı/Başarısız gibi sadece iki sonucu olan durumlarda

Örnek: Web sitenizde yeni bir tasarımı test ediyorsunuz. Yeni tasarım gerçekten daha fazla tıklama alıyor mu?

Yorumlama: 0.05 ten küçükse bu fark tesadüf değil diyebiliriz

### **PSI (Population Stability Index)**

Ne zaman kullanılır: Zaman içinde değişimi ölçmek istediğimizde

Örnek: Müşteri davranışları geçen aya göre değişti mi?

Yorumlama:
    
    - 0.1 den küçük: Her şey normal
    - 0.1 - 0.2 arası: Biraz değişim var
    - 0.2 den büyük: Önemli bir değişim var, dikkat etmek gerek

### **AUC-ROC**

Ne zaman kullanılır: Tahmin modelimiz ne kadar iyi çalışıyor anlamak istediğimizde

Örnek: Spam e-posta tespit modelimiz ne kadar başarılı?

Yorumlama:

    - 0.5: Model hiç iyi değil (Yazı-Tura atmak gibi)
    - 0.7 - 0.8: İyi
    - 0.9 dan büyük: Çok iyi
    - 1.0: Mükemmel (ama gerçek hayatta pek rastlanmaz)

### **RMSE ve MAE**

Ne zaman kullanılır: Tahminlerimizin ne kadar şaştığını ölçmek istediğimizde

Örnek: Ev fiyatı tahmin modelimiz gerçek fiyatlardan ortalama ne kadar sapıyor?

Yorumlama: Küçük değerler daha iyidir (tahminler gerçeğe daha yakın demektir)

### **F1-Score**

Ne zaman kullanılır: Doğru ve yanlış tahminlerimizi tek bir sayıyla özetlemek istediğimizde

Örnek: Hastalık teşhis modelimizin genel başarısını ölçmek istiyoruz

Yorumlama: 
    
    - 0: En kötü
    - 1: En iyi
    - Genelde 0.7 den büyük değerler iyi kabul edilir

### **Chi-Square Testi**

- Ne zaman kullanılır: İki kategori arasında bir ilişki var mı yok mu anlamak istediğimizde

- Örnek: Eğitim seviyesi ile gelir düzeyi arasında bir ilişki var mı?

Yorumlama: p-değeri 0.05 ten küçükse aralarında bir ilişki var diyebiliriz

### **Balanced Accuracy**

Ne zaman kullanılır: Dengesiz veri setlerinde kullanılır

Her sınıfın doğruluk oranlarını eşit ağırlıklandırır

Yorumlama: Accuracy den düşükse dengesizlik vardır

### **Matthews Correlation**

Ne zaman kullanılır: Dengesiz veri setlerinde

Confusion matrix'in tüm değerlerini kullanır

Yorumlama: -1, +1 arası değer alır. >0.4 iyi değer kabul edilir

En güvenilir binary sınıflandırma metriklerinden

### **Log Loss**

Ne zaman kullanılır: Olasılık tahminlerinin önemli olduğu durumlarda

Tahmin keskinliğini ölçer

Yorumlama: 0'a yakın olması daha iyi

Model kalibrasyonunu değerlendirir

### **Brier Score**

Ne zaman kullanılır: Olasılık kalibrasyonu önemliyse

Tahmin olasılıklarının doğruluğunu ölçer

Yorumlama: 0-1 arası değer alabilir, 0'a yakın olması iyi
