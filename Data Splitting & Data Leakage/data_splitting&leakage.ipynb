## Data Splitting

Cross Validation (Çapraz Doğrulama)

Eğitim verilerinin bir alt kümesini dışarıda tutarak test hatasını tahmin eden yöntemdir.

Validation Set: Verileri eğitim seti ve doğrulama seti olarak ayırın. Modeli eğitim üzerine eğitin ve doğrulamayı kullanarak test hatasını tahmin edin. Örneğin 80-20 bölünmüş

k-Fold CV: Verileri yaklaşık olarak eşit büyüklükteki k gruba rastgele bölün. İlk kat doğrulama, geri kalanı eğitim olarak kullanılır. Ardından k kez tekrarlayın ve k tahminlerinin ortalamasını bulun.

Leave-One-Out CV: Verileri eğitim seti ve doğrulama seti olarak ayırın, ancak doğrulama seti 1 gözlemden oluşur. Ardından, tüm gözlemler doğrulama olarak kullanılana kadar n-1 kez tekrarlayın. Test hatası, bu n test hatası tahminlerinin ortalamasıdır.

## Data Leakage

- Eğitim verilerinin, gerçek dünyada mevcut olmayacak hedef bilgilerini içermesi
- Modelin eğitim ve doğrulama aşamalarında yanıltıcı yüksek performans göstermesi
- Gerçek dünyada kullanıldığında modelin başarısız olması

Veri sızıntısının iki ana türü
- Hedef sızıntısı
- Eğitim - test kontaminasyonu

# Hedef Sızıntısı
- Tahmin yapılacağı zaman mevcut olmayacak verilerin kullanılması
- Örnek: Zatürre tahmini yaparken antibiyotik kullanımı bilgisinin kullanılması
- Çözüm: Hedef değer belirlendikten sonra güncellenen değişkenlerin çıkarılması

# Eğitim - Test kontaminasyonu
- Eğitim ve doğrulama verilerinin dikkatlı ayrılmaması
- Ön işleme adımlarının tüm veri seti üzerinde yapılması
- Çözüm: Scikit-learn pipeline'larının kullanılması, çapraz doğrulama

# Veri Sızıntısını Tespit Etme
- Şüpheli derecede yüksek model performansı
- Veri setinin dikkatli incelenmesi
- Temel veri karşılaştırmaları yapma

# Veri sızıntısı önleme stratejileri
- Veri hazırlama sürecinin dikkatli incelenmesi
- Zaman bazlı veri bölünmesi kullanma
- Scikit-learn pipeline'larını kullanma
- Çapraz doğrulama sırasında ön işleme adımlarını pipeline içinde tutma

# Sonuç
- Veri sızıntısı, milyon dolarlık hatalara yol açabilir
- Dikkat veri hazırlığı ve model değerlendirilmesi çok önemlidir
- Sağduyu, dikkat ve veri keşfi, hedef sızıntısını belirlemeye yardımcı olur
