### Data Warehouse (Veri Ambarı)

Amacı:

- Farklı kaynaklardan gelen verileri tek bir yerde toplamak
- Veriye dayalı karar vermeyi kolaylaştırmak

Gerçek hayattan örnek:

Bir süpermarket zinciri düşünelim. Her şubenin satışları, stokları, müşteri bilgileri gibi veriler vardır. Bu veriler, merkezi bir depoda toplanır, düzenlenir ve analiz için saklanır. İşte bu merkezi depoya Data Warehouse veya Veri Ambarı denir

Benzerlik:

- Kütüphane: Farklı raflarda, konulara göre düzenlenmiş kitaplar gibi; veri ambarında da farklı kaynaklardan gelen veriler, analiz ve raporlama için düzenlenmiştir.

### Hadoop Ne işe Yarar?

Hadoop, devasa miktarda veriyi, birçok bilgisayar arasında dağıtarak işleyen bir sistemdir

Gerçek hayattan örnek:

Büyük bir fabrika düşünelim. Fabrika, iş yükünü tek bir makinede değil, birçok işçiye dağıtarak yapıyor. İşin her bir parçası farklı çalışanlar tarafından yapılır ve sonunda sonuçlar birleştirilir

Benzerlik:

- Tıpkı fabrikanın işlerini birçok işçiye dağıtması gibi, Hadoop da veriyi küçük parçalara böler ve bu parçaları paralel olarak işler

### Hive Neden Önemli?

Hive, Hadoop'un karmaşık verilerine SQL benzeri sorgular yazmaya olanak tanıyan bir sistemdir

Kütüphanede aradığımız konuyla ilgili kitapları bulmamız için bir rehber olarak düşünebiliriz

### Parquet Neden Tercih Edilir?

Parquet, sütun bazlı bir dosya formatıdır, böylece sadece gerekli sütunlara erişerek veriyi daha hızlı işleyebiliriz

Gerçek hayattan örnek:

geniş sütunlar ve satırlardan oluşan devasa bir tablomuz olduğunu düşünelim. İhtiyacımız olan sadece belirli sütunlarsa, tüm tabloyu okumak yerine sadece o sütunları çıkarmak isteriz, Veriler sütunlar halinde saklandığı için parquet, gereksiz tekrarları önleyerek depolamayı optimize eder.


Özet:

- Data Warehouse: Farklı kaynaklardan gelen verilerin merkezi bir yerde toplanıp düzenlendiği depo
- Hadoop: Büyük veriyi, birbirine bağlı birçok bilgisayar arasında bölerek işleyebilen sistem
- Hive: Hadoop üzerinde SQL benzeri sorgularla verileri kolayca analiz etmemizi sağlayan araç
- Parquet: Sütun bazlı depolama formatı ile veri okuma ve işleme performansını artıran sistem

### Data Lake (Veri Gölü) Nedir?

Data Lake, ham verilerin doğal/işlenmemiş formatta depolandığı büyük bir havuzdur.

Gerçek hayattan örnek:
Bir göl düşünelim - içinde her türden su birikir. Benzer şekilde, 
veri gölü de her formattaki veriyi olduğu gibi saklar.

Önemli özellikleri:
- Ham verileri orijinal formatında saklar
- Esnek yapısı sayesinde her türlü veriyi alabilir
- Veri bilimciler için zengin bir kaynak oluşturur


Data Lake ve Data Warehouse arasındaki temel fark: Data Warehouse yapılandırılmış ve işlenmiş verileri saklarken, Data Lake ham verileri orijinal formatında saklar. Bu nedenle Data Lake daha esnek bir yapıya sahiptir ve veri bilimciler için daha zengin bir kaynak oluşturur.
