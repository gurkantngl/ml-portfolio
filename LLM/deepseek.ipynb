DeepSeek Nedir?

Aralık 2023'te ortaya çıkan ve kısa sürede OpenAI, Google ve Anthropic gibi devlerle rekabet etmeye başlayan bir Çinli yapay zeka şirketidir

En yeni modelleri R1, önde gelen modellerle rekabet edebiliyor. 671 miilyar parametreye sahip ve ileri düzey matematik testlerinde %90,8 başarı oranı yakalayarak birçok ölçütte GPT-4 ile aynı performansı gösteriyor.

#### DeepSeek'in teknik yenilikleri:

- Mixture of Experts (MoE) yaklaşımı: DeepSeekMoE, modelin sadece ilgili uzmanlık alanlarını etkinleştirerek gereksiz hesaplama yükünü azaltıyor. Bu, modelin daha verimli çalışmasını sağlıyor

- Multi-Head Latent Attention (MLA): DeepSeekMLA, bellek kullanımını optimize ederek modelin daha uzun bağlamları işlemesine olanak tanıyor

- Reinforcement Learning (Pekiştirmeli Öğrenmeli): R1 modeli, insan müdahalesi olmadan, sadece pekiştirmeli öğrenme kullanarak kendi kendine gelişmiş akıl yürütme yetenekleri geliştirdi

#### Maliyet ve Kaynak Verimliliği

V3 modelini yaklaşık 2.788 milyon GPU saati kullanarak ve sadece 5.576 milyon dolarlık bir maliyetle eğittiğini belirtiyor. Bu, diğer büyük yapay zeka modellerine kıyasla oldukça düşük bir maliyet ve yüksek bir verimlilik anlamına geliyor.

#### Reasoning

IBM üyesi Kush Varshney, "Gerçekten etkileyici olan şey DeepSeek modellerinin akıl yürütme yeteneğidir" diyor. Akıl yürütme modelleri esasen kendilerini doğrular veya denetler. Artık bu modellere bilgelik katmaya başlıyoruz ve bu çok büyük bir adım

DeepSeek R1, düşünce zinciri akıl yürütmeyi, otonom bir etkenin deneme yanılma yoluyla ve bir insan kullanıcıdan herhangi bir talimat almadan bir görevi gerçekleştirmeyi öğrendiği takviyeli öğrenme ile birleştirir

#### Açık kaynak için yeni bir dönem mi?

Deepseek, açık kaynak bir model olarak kullanıcılara model üzerinde değişiklik yapma, geliştirme ve özelleştirme özgürlüğü sunuyor. Bu durum, ChatGPT ve Claude gibi kapalı kaynak modellerden ayrılıyor. Ayrıca, açık kaynaklı yapay zekanın düşük maliyetle, ticari modellerle rekabet edebileceğini kanıtlıyor.

#### Sonuç

DeepSeek, sınırsız kaynaklara veya en gelişmiş donanuma ihtiyaç duymadan rekabetçi bir yapay zeka modeli geliştirilebileceğini kanıtlıyor.
