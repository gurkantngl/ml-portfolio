- Sapma, bir ML modelinin zaman içinde performansının düşmesine neden olan değişimlerdir

Neden Oluşur?
- Eğitim verilerinin zamanla eskimesi
- Gerçek dünya koşullarının değişmesi
- Veri dağılımındaki değişiklikler

Basit Bir Örnekle
* Başlangıç: Model ev fiyatlarını tahmin etmek için 2020 verileriyle eğitildi
* Sorun: 2024'te ekonomik koşullar değişti
* Sonuç: Model artık doğru tahminler yapamıyor


Sapmanın Etkileri
- Model doğruluğunda düşüş
- Beklenmeyen performans sorunları
- Yanlış tahminler ve kararlar


Çözüm Yaklaşımları
1. Düzenli model güncellemeleri
2. Sürekli performans izleme
3. Yeni verilerle yeniden eğitim
4. Sapma tespiti için otomatik sistemler


Önemli Not
- Sapma kaçınılmazdır
- Proaktif izleme ve müdahale gerektirir
- Modelin sürdürülebilirliği için kritik öneme sahiptir

## Sapma Türleri

1. Kavram Sapması (Concept Drift)
Tanım

- Model görevinin zamanla değişmesi
- Tahmin edilecek hedef değişkenin değişimi

Örnek: Spam Tespiti

- Başlangıç: Model klasik spam maillerini tanır
- Değişim: Spam türleri değişir
- Sonuç: Model yeni spam türlerini tespit edemez

Alt Türleri

1. Ani Sapma

- Ani ve keskin değişimler
- Hızlı performans düşüşü

2. Kademeli Sapma

- Yavaş ve sürekli değişim
- Zamanla artan performans kaybı

3. Artımlı Sapma

- Adım adım değişim
- Belirli aralıklarla görülen düşüş

4.Tekrarlayan Sapmalar

- Döngüsel değişimler
- Mevsimsel etkiler



2. Veri Sapması (Data Drift)

Tanım

- Giriş verilerinin dağılımında değişim
- Eşdeğişken sapması olarak da bilinir

Örnek: Müşteri Analizi

- Başlangıç: Belirli yaş/gelir dağılımı
- Değişim: Müşteri demografisi değişir
- Sonuç: Tahminler tutarsızlaşır

3. Çözüm Stratejileri

- Sürekli performans izleme
- Düzenli model güncellemeleri
- Dayanıklı model tasarımı
- Erken uyarı sistemleri

4. Önemli Noktalar

- Her iki sapma türü de ciddi etkiye sahip
- Proaktif yaklaşım gerekli
- Sürekli izleme şart
- Düzenli müdahale önemli

## Sapma Tespit Yöntemleri

1. Model Tabanlı Yaklaşım

Temel Özellikler

- Giriş verilerindeki değişimleri izler
- Gerçek zamanlı kontrol sağlar
- Model performansını sürekli değerlendirir

2. İstatistiksel Testler

A. Sıralı Analiz Yöntemleri

- Veri dizilerini inceler
- Zaman içindeki değişimleri takip eder
- Anormal sapmaları tespit eder

B. Özel Sapma Tespit Modelleri

- Sapma tespiti için özelleştirilmiş
- Belirli veri türlerine odaklanır
- Yüksek hassasiyetle çalışır

C. Zaman Dağılımı Yöntemi
Kullanılan Metrikler:

- Popülasyon Kararlılık İndeksi

    - Veri dağılımının stabilitesini ölçer

- KL Sapması

    - Dağılımlar arası farklılıkları ölçer

- JS Sapması

    - KL sapmasının simetrik versiyonu

- KS Testi

    - Dağılımların benzerliğini test eder
    
- Wasserstein Metriği

    Dağılımlar arası mesafeyi ölçer

3. Önemli Noktalar

- Düzenli kontrol gerektirir
- Birden fazla yöntem kullanılmalı
- Erken uyarı sistemi kurulmalı
- Sonuçlar sürekli izlenmeli

Sapma Tespit Algoritmaları ve Yöntemleri

1. Kolmogorov-Smirnov (K-S) Testi

Temel Özellikleri

- Parametrik olmayan istatistiksel test
- İki veri seti karşılaştırması
- Dağılım benzerliği analizi

Çalışma Prensibi

- Sıfır hipotezi: Dağılımlar aynıdır
- Hipotez reddi: Sapma var demektir
- Veri setleri arası karşılaştırma yapar

2. Popülasyon Kararlılık İndeksi (PSI)

Ana Özellikler

- Kategorik değişken analizi
- Zaman içi değişim ölçümü
- Risk skorlama aracı

Kullanım Alanları

1. Model performans izleme
2. Dağılım değişikliği tespiti
3. Bağımlı/bağımsız değişken analizi

Değerlendirme

- Yüksek PSI = Önemli dağılım farkı
- Düşük PSI = Stabil dağılım
- Model yenileme kararı için kullanılır

3. Page-Hinkley Yöntemi

Temel Bileşenler

1. Eşik değeri belirleme
2. Karar fonksiyonu tanımlama
3. Ortalama değişim takibi

Uygulama Adımları

- Sürekli veri izleme
- Ortalama hesaplama
- Değişim tespiti
- Alarm mekanizması

Avantajları

- Basit ve etkili
- Küçük değişimleri tespit edebilir
- Gerçek zamanlı izleme

4. MLOps Entegrasyonu

Makine öğrenmesi modellerini sapma açısından izlemek, MLOps adı verilen daha geniş bir alanın sadece bir yönüdür. MLOps kavramlarını anlamak, makine öğrenmesi modellerini yerel bir not defterinden üretimde çalışan bir modele dönüştürmek isteyen her veri bilimcisi, mühendis veya lider için önemlidir.

5. Sonuç Önerileri

1. Çoklu yöntem kullanımı
2. Düzenli kontrol mekanizmaları
3. Otomatik müdahale sistemleri
4. Sürekli iyileştirme döngüsü
