### Makine Öğrenimi Sistemlerinde "Fairness" Kavramı

"Fairness" kavramı, Yapay Zeka (AI) ve Makine Öğrenimi (ML) alanında yaygın olarak kullanılan bir terimdir.

Ancak, "fairness" kavramının ne anlama geldiği ve nasıl uygulanacağı konusu hala tartışmalıdır.

#### Fairness, farklı disiplinlerde farklı tanımlara sahiptir:

Hukuk: ayrımcılık ve kötü muameleyi önleme

Sosyal bilim: sosyal ilişkiler, güç dinamikleri, kurumlar ve pazarlar

Kantitatif alanlar: matematiksel problemler, eşit veya adil dağılım

Felsefe: ahlaki doğruluk Disiplinlerde Adalet Anlayışı

### Makine Öğrenimi Sistemlerinde Fairness

ML sistemlerinde fairness, genellikle nicel bir perspektifle ele alınır.

ML araştırmacıları ve uygulayıcıları, optimal bir ML modeli oluşturmak için fairness kısıtlamalarını kullanır.

Fairness kısıtlamaları, yasalar, sosyal bilim ve felsefe perspektiflerinden bilgi alabilir.

### Fairness İle İlgili Sorunlar

Fairness kavramının farklı tanımları ve uygulamaları:

ML sistemlerinde fairnessin nasıl sağlanacağı konusu hala tartışmalıdır

ML sistemleri, toplumun mevcut eşitsizliklerini yansıtabilir ve hatta artırabilir

### COMPAS Algoritması Örneği

- COMPAS algoritması, mahkemelerde kullanılan bir risk değerlendirme aracıdır

- Algoritma, nicel bir yaklaşımı kullanarak fairnessi sağlamaya çalışır

- Ancak, algoritma, siyah ve beyaz tutuklular arasında farklı sonuçlar üretir
