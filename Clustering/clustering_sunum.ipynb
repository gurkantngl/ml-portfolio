# Case Nedir?

HELP - International
 - Ülke Sınıflandırma ve Yardım Önceliklendirme Projesi

Proje Özeti

HELP International, yoksullukla mücadele eden ve gelişmekte olan ülkelere insani yardım sağlayan uluslararası bir sivil toplum kuruluşudur. Kuruluş, afet dönemlerinde temel ihtiyaçların karşılanması ve yardım faaliyetlerinin yürütülmesi konusunda aktif rol oynamaktadır.

Problem Tanımı

CEO'nun karşı karşıya olduğu temel soru: "Bu yardım fonunu hangi ülkelere, nasıl dağıtmalıyız?"

Proje Hedefi

- Ülkeleri çeşitli gelişmişlik göstergelerine göre sınıflandırma
- Yardıma en çok ihtiyaç duyan ülkeleri belirlemek
- Kaynakların etkili kullanımı için önceliklendirme yapmak

Değerlendirme Kriterleri

- Sosyo-ekonomik göstergeler
- Sağlık göstergeleri
- Yaşam kalitesi endeksleri
- Temel ihtiyaçlara erişim durumu

Beklenen Sonuç

Veri analizi sonucunda, yardım fonunun hangi ülkelere öncelikli olarak aktarılması gerektiğine dair net bir yol haritası oluşturulacaktır. Bu sayede, sınırlı kaynakların en yüksek etkiyi yaratacak şekilde kullanılması sağlanacaktır.

child_mort -> 5 yaş altı çocuk ölümleri, 1000 canlı doğumunda

exports -> Kişi başına mal ve hizmet ihracatı. Kişi başına GSYİH'nın % si verilmektedir

health -> Kişi başına toplam sağlık harcamaları. Kişi başına GSYİH'nın %'si olarak verilmiştir

imports -> Kişi başına mal ve hizmet ithalatı. Kişi başına GSYİH'nın %si olarak verilmektedir

income -> Kişi başına net gelir

inflation -> Toplam GSYİH'nın yıllık büyüme oranının ölçümü

life_expec -> Mevcut ölüm oranlarının aynı kalması durumunda, yeni doğan bir çocuğun ortalama yaşam süresi

total_fer -> Her kadının doğurganlık yaş oranları aynı kalırsa, doğacak çocuk sayısı

gdpp -> Kişi başına GSYİH. Toplam GSYİH'nın toplam nüfusa bölünmesiyle hesaplanır

## Clustering (Kümeleme)
- Kümeleme, benzer özelliklere sahip verileri gruplandırma işlemidir. Örneğin;

    - Benzer alışveriş alışkanlıkları olan müşterileri gruplandırma
    - Benzer özellikteki resimleri gruplandırma
    - Benzer davranış gösteren kullanıcıları gruplandırma

* Çalışma Mantığı:

    - Veriler arasındaki benzerlikleri ölçer
    - Benzer verileri aynı kümeye atar
    - Farklı verileri farklı kümelere ayırır

- Kullanım Alanları:
    
    - Müşteri Segmentasyonu
    - Görüntü sıkıştırma
    - Anomali Tespiti
    - Öneri sistemleri
    - Pazar araştırması

- En yaygın kümeleme algoritmaları:
    - K-Means: En popüler ve basit kümeleme algoritması
    - Hierarchical Clustering: Hiyerarşik Kümeleme
    - DBSCAN: Yoğunluk tabanlı kümeleme

- Kümelemenin Avantajları:
    - Veri içindeki gizli yapıları keşfeder
    - Otomatik gruplandırma sağlar
    - Veri analizi için değerli içgörüler sunar

- Dikkat edilmesi gerekenler:
    - Küme sayısını doğru belirlemek önemlidir
    - Veri ön işleme gerekebilir
    - Sonuçların yorumlanması uzmanlık gerektirebilir


Kümeleme, makine öğrenmesinin "gözetimsiz öğrenme" (unsupervised learning) kategorisine girer, yani verileri etiketlere ihtiyaç duymadan gruplandırabilir. Bu özelliği sayesinde, önceden bilinmeyen örüntüleri keşfetmekte çok başarılıdır.

# K-Means (K-Ortalama)
Nasıl çalışır?

    - Kaç küme olacağı kullanıcıdan parametre olarak seçilir
    (Kaç segment olacağı ile ilgili fikrimizin olacağı kabul edilir)
    - Rastgele olarak k merkez noktası seçilir
    - Her veri örneği en yakın merkez noktasına göre ilgili kümeye atanır
    - Her küme için yeni merkez noktaları hesaplanarak merkez noktaları kaydırılır

Çalışma Mantığı (Adım adım)

    * Başlangıç:
        - K adet rastgele merkez nokta seçilir
    * Tekrarlanan Adımlar:
        - Her veri noktası, en yakın merkeze atanır
        - Her kümenin yeni merkezi, o kümedeki noktaların ortalaması alınarak güncellenir
        - Bu işlem merkezler sabitlenene kadar devam eder
    
Avantajları:

    - Anlaşılması ve uygulaması kolay
    - Büyük veri setlerinde hızlı çalışır
    - Kümeleri eşit boyutlarda oluşturma eğilimindedir

Dezavantajları:

    - K değerini önceden belirlememiz gerekir
    - Başlangıç merkezlerine duyarlıdır
    - Dairesel olmayan kümelerde zorlanabilir
    - Aykırı değerlere karşı hassastır

- Kümeleme algoritmasının başarısı nasıl ölçülür?

    Bölütler içindeki veriler arasındaki mesafe minimum, bölütler arasındaki mesafe maksimum olmalı

## Dirsek yöntemi(Elbow Methodu)

- Temel Amaç:

    - En uygun küme sayısını (K değerini) belirlemek
    - Veri setini gereksiz yere fazla kümeye bölmemek
    - Yetersz kümeleme yapmamak

- Çalışma Mantığı:

    - Her K değeri için "inertia" değeri hesaplanır (veri noktalarının kendi küme merkezlerine olan uzaklarının karelerinin toplamıdır, yani verinin kümeler içinde ne kadar dağınık olduğunu gösteren bir ölçümdür)
    - K arttıkça bu değer doğal olarak azalır çünkü:
        - Daha fazla küme = Noktalara daha yakın mesafeler
        - K = veri sayısı olduğunda inertia = 0 olur
    
- Neden dirsek denir?

    - Grafik insan koluna benzer bir şekil oluşturur
    - Dirsek kısmındaki bükülme noktası, optimal K değerini gösterir
    - Bu noktadan sonra K'yı artırmak çok az fayda sağlar

- Dikkat edilmesi gerekenler:
    - Bazen net bir dirsek noktası olmayabilir
    - Veri setinin yapısına göre farklı sonuçlar alınabilir
    - Diğer metriklerle birlikte kullanılması önerilir (silhoutte score gibi)

* Dirsek yöntemi, K-Means için optimal K değerini belirlemede en popüler yöntemlerden biridir. Ancak her zaman tek başına yeterli olmayabilir. Veri setinizin yapısına göre silhouette analizi gibi diğer yöntemlerle birlikte kullanmanız daha sağlıklı sonuçlar verebilir.

## Silhoutte (Siluet) Analizi

Silhoutte analizi, kümeleme kalitesini ölçen bir yöntemdir.

- Temel Mantık:
    - Her veri noktası için -1 ile +1 arasında bir skor hesaplar
    - +1'e yakın değerler iyi kümelemeyi gösterir
    - -1'e yakın değerler yanlış kümelemeyi gösterir

Silhouette Score = (b - a) / max(a, b)

Burada:
a = Noktanın kendi kümesindeki diğer noktalara ortalama uzaklığı
b = Noktanın en yakın komşu kümedeki noktalara ortalama uzaklığı

- Skorların Yorumlanması

    * 0.7 - 1.0: Güçlü kümeleme yapısı
    * 0.5 - 0.7: Orta düzeyde kümeleme yapısı
    * 0.25 - 0.5: Zayıf kümeleme yapısı
    * < 0.25: Kötü niteleme yapısı

- Avantajları:

    - Küme sayısını belirlemede etkili
    - Kümeleme kalitesini objektif olarak ölçer
    - Dirsek yöntemiyle birlikte kullanılabilir

- Dezavantajları:
    
    - Hesaplama süresi büyük veri setlerinde uzun olabilir
    - En az 2 küme gerektirir
    - Bazı durumlarda yanıltıcı sonuçlar verebilir

* Silhoutte analizi, özellikle dirsek yöntemiyle birlikte kullanıldığında, optimal küme sayısını belirlemede çok etkili bir araçtır

# DBSCAN

- Yoğunluk tabanlı bir kümeleme algoritmasıdır
- K-Means'in aksine küme sayısını önceden belirtmeye gerek yoktur
- Gürültülü verileri (noise/outliers) tespit edebilir
- Düzensiz şekilli kümeleri bulabilir

Temel Parametreleri:

    - eps (ε): Bir noktanın etrafında komşularını aramak için kullanılan maksimum mesafe (yarıçap) değeridir.
    - min_samples: Bir noktanın "çekirdek nokta" olarak kabul edilebilmesi için eps yarıçapı içinde bulunması gereken minimum nokta sayısıdır.

Çalışma Mantığı:

    - Her nokta için 3 durum vardır:
        - Çekirdek Nokta (Core Point)
        - Sınır Nokta (Border Point)
        - Gürültü Nokta (Noise Point)

Nokta Türleri:

    - Çekirdek Nokta: eps yarıçapında min_samples kadar komşusu olan nokta
    - Sınır Nokta: Çekirdek noktanın eps yarıçapında olan ama kendisi çekirdek olmayan nokta
    - Gürültü Nokta: Ne çekirdek ne de sınır nokta olan noktalar (-1 ile etiketlenir)

Avantajları:

    - Küme sayısını önceden belirtmeye gerek yok
    - Aykırı değerleri tespit edebilir
    - Düzensiz şekilli kümeleri bulabilir
    - Gürültülü verilerde iyi çalışır

Dezavantajları:
    
    - eps ve min_samples parametrelerini belirlemek zor olabilir
    - Yoğunluğu çok farklı kümelerde zorlanabilir
    - Yüksek boyutlu verilerde mesade hesabı problemli olabilir

Gerçek dünya uygulamaları:
    
    - Gürültü tespiti
    - Anomali tespiti
    - Mekansal veri analizi
    - Görüntü segmentasyonu

K - Means ile Temel Farklar:

    - Küme sayısı önceden belirlenmez
    - Düzensiz şekilli kümeleri bulabilir
    - Gürültü noktalarını tespit edebilir
    - Tüm noktaları bir kümeye atamak zorunda değil

DBSCAN, özellikle gürültülü verilerde ve düzensiz şekilli kümelerde K-Means'e göre daha iyi sonuçlar verir. Ancak parametre seçimi kritiktir ve veri setinin yapısına göre bu parametreleri doğru belirlemek önemlidir

# Hierarchical Clustering (Hiyerarşik Kümeleme):

- Verileri hiyerarşik bir ağaç yapısında (dendogram) kümelere ayırır
- İki temel yaklaşımı vardır:
    - Agglomerative (Birleştirici): Aşağıdan Yukarıya
    - Divisie (Bölücü): Yukarıdan Aşağıya

Algoritma Adımları (Agglomerative)

    - Her veri tek bir küme / bölüt ile başlar
    - En yakın ikişer komşuyu alıp ikişerli küme/bölüt oluşturur
    - En yakın iki kümeyi alıp yeni bir bölüt oluşturur
    - Bir önceki adım, tek bir bölüt/küme olana kadar devam ederL

- En yaygın kullanılan yöntem (Agglomerative)

    - Her nokta başlangıçta ayrı bir küme
    - En yakın iki küme birleştirilir
    - İstenilen küme sayısına ulaşana kadar devam eder

- Avantajları:

    - Küme sayısını önceden belirtmek zorunlu değil
    - Hiyerarşik yapıyı görselleştirebilme (dendogram)
    - Farklı seviyelerde kümeleme imkanı
    - Sezgisel ve anlaşılır

- Dezavantajları:

    - Büyük veri setlerinde yavaş (o(n^3)) karmaşıklık
    - Bellek kullanımı yüksek
    - Gürültüye karşı hassas
    - Birleştirmeler geri alınamaz
