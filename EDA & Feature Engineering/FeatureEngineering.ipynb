## Feature Engineering (Özellik Mühendisliği)

- Feature Engineering, makine öğrenimi algoritmalarının daha iyi performans göstermesine yardımcı olan özellikler veya girdi değişkenleri oluşturmak için alan bilgisini kullanma sürecidir.

Doğru yapıldığında, modellerinizin tahmin gücünü artırmanıza yardımcı olabilir.

Feature Engineering bilimden çok bir sanattır. FE, iyi bir model oluşturmanın en önemli adımlarından biridir.


"Özellikleri bulmak zor, zaman alıcı, uzman bilgisi gerektiriyor. 'Uygulamalı makine öğrenimi' temelde özellik mühendisliğidir"


- Sürekli Veri

Raw measures: henüz dönüştürülmemiş veriler (kaynaktan alınan veri)

Rounding: bazen kesinlik gürültülüdür, en yakın tamsayıya yuvarlama, ondalık vb.

Scaling: log, z-skoru, minmax ölçeği

Imputation: ortalama, medyan, model çıktısı vb. kullanarak eksik değerleri doldurma.

Binning: Sayısal özellikleri kategorik olanlara dönüştürme, ör. 1-10 arasındaki değerler A'ya, 10-20 arasındaki değerler B'ye aittir

Interactions: Özellikler arasındaki etkileşimler. ör. çıkarma, toplama, çarpma, istatistiksel test, log/güç dönüşümü (çarpık dağılımları daha normal hale getirmeye yardımcı olur)

Dimensionality Reduction: PCA, kümeleme, faktör analizi vb. (yüksek boyutlu veri setlerini daha düşük boyutlu bir forma dönüştürme işlemidir)


- Ayrık Veri

Encoding: Bazı ML algoritmaları kategorik veriler üzerinde çalışamadığı için kategorik verilere veya faktörlere dönüştürmemiz gerekiyor.

Label Encoding:
    - Her kategorik değere benzersiz bir sayısal değer atanır.
    - Örnek: ['R','G','B'] -> [0, 1, 2]
    - Avantajı: Basit ve az yer kaplar
    - Dezavantajı: Sayısal değerler arasında yanlış bir sıralama ilişkisi oluşturabilir

One-Hot Encoding:
    - Her kategori için yeni bir sütun oluşturulur ve ilgili kategori 1, diğerleri 0 olarak işaretlenir.
    - Örnek: ['R','G','B'] -> [[1,0,0],[0,1,0],[0,0,1]]
    - Avantajı: Kategoriler arasında yanlış sıralama ilişkisi oluşturmaz.
    - Dezavantajı: Çok fazla kategori olduğunda boyut artışına neden olabilir.


```python
df = pd.read_csv("column_3C_weka.csv")
```

```python
def grab_col_names(dataframe, cat_th=10, car_th=20):

    cat_cols = [col for col in dataframe.columns if dataframe[col].dtypes == "O"]
    num_but_cat = [col for col in dataframe.columns if dataframe[col].nunique() < cat_th and dataframe[col].dtypes != "O"]
    cat_but_car = [col for col in dataframe.columns if dataframe[col].nunique() > car_th and dataframe[col].dtypes == "O"]
    cat_cols = cat_cols + num_but_cat
    cat_cols = [col for col in cat_cols if col not in cat_but_car]
    num_cols = [col for col in dataframe.columns if dataframe[col].dtypes != "O"]
    print(f"Observations: {dataframe.shape[0]}")
    print(f"Variables: {dataframe.shape[1]}")
    print(f"cat_cols: {len(cat_cols)}")
    print(f"num_cols: {len(num_cols)}")
    print(f"cat_but_car: {len(cat_but_car)}")
    print(f"num_but_cat: {len(num_but_cat)}")
    return cat_cols, num_cols, cat_but_car
```

```python
cat_cols, num_cols, cat_but_car = grab_col_names(df)
```
