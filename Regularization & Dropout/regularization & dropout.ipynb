# Regularization

- Regularization, overfittingi önleyen bir tekniktir.
- Modele, bir çeşit ceza vererek, çok karmaşık çözümler üretmesini engeller.
- Böylece model daha genelleştirilebilir hale gelir ve yeni veriler üzerinde daha iyi performans gösterir.

En yaygın regularization teknikleri:

- L1 Regularization
- L2 Regularization
- Dropout
- Early Stopping

## L1 Regularization

- L1 Regularization, modeldeki bazı özelliklerin (feature) ağırlıklarını tam olarak 0'a indirir.
- Yani önemsiz özellikleri otomatik elemek için kullanılır.
- Formül: Loss = Original_Loss + λ * Σ|weights|
    λ (lambda): regularization şiddetini belirleyen parametre
    |weights|: ağırlıkların mutlak değeri

Avantajları:

    - Feature Selection yapar
    - Model karmaşıklığını azaltır
    - Daha seyrek bir model oluşturur
    - Modeli daha yorumlanabilir hale getirir

Ne zaman kullanılır:

    - Veri setinizde çok fazla özellik varsa
    - Hangi özelliklerin önemli olduğunu bulmak istiyorsanız
    - Model boyutunu küçültmek istiyorsanız

## L2 Regularization

- L2 Regularization, ağırlıkları sıfıra indirmek yerine küçültmeye çalışır
- Büyük ağırlıkları cezalandırır ama tamamen sıfırlamaz
- Formülü: Loss = Original_Loss + λ * Σ(weights²)
    λ (lambda): regularization şiddetini belirleyen parametre
    weights²: ağırlıkların karesi

Avantajları:
- Ağırlıkları dengeli bir şekilde küçültür
- L1'e göre daha yumuşak bir düzenleme sağlar
- Özellikler arasında yüksek korelasyon olduğunda iyi çalışır
- Gradyan hesaplama L1'e göre daha kolaydır


Ne zaman kullanılır:
- Özelliklerinizin hepsinin bir miktar önemli olduğunu düşünüyorsanız
- Özellikleri tamamen elemek yerinde etkilerini azaltmak istiyorsanız

L1 ve L2 arasındaki fark:
- L1: Bazı ağırlıkları tamamen 0 yapar
- L2: Ağırlıkları küçültür ama 0 yapmaz

## Dropout

- Eğitim sırasında sinir ağındaki nöronların rastgele bir kısmını geçici olarak devre dışı bırakır
- Örneğin 0.5 dropout oranı, her eğitim adımında nöronların %50'sinin rastgele kapatılacağı anlamına gelir
- Test aşamasında tüm nöronlar aktif olur


Avantajları:

- Overfitting'i önler
- Nöronların birbirine aşırı bağımlı olmasını engeller
- Model daha dayanıklı hale gelir
- Farklı nöron kombinasyonlarıyla çalışmayı öğrenir


Ne zaman kullanılır:

- Derin sinir ağlarında
- Overfitting problemi yaşandığında
- Büyük ve karmaşık modellerde
- Özellikle görüntü işleme ve doğal dil işleme görevlerinde

# Early Stopping

- Model eğitimi sırasında validation hatasını sürekli takip eder
- Validation hatası artmaya başladığında eğitimi durdurur
- En iyi performansı gösteren model ağırlıklarını kaydeder


Avantajları:

- Overfitting'i önler
- Eğitim süresini kısaltır
- Otomatik olarak en iyi modeli seçer
- Hesaplama kaynaklarından tasarruf sağlar

Ne zaman kullanılır:
- Derin öğrenme modellerinde
- Eğitim süresini optimize etmek istediğinizde
- Overfitting riski olan durumlarda
- Büyük veri setlerinde

* Bu şekilde model, gereksiz yere uzun süre eğitilmek yerine, en optimal performans noktasında durdurulur

Precision

- Pozitif dediğimiz sonuçların ne kadar gerçekten pozitif?
- Formül: TP / (TP + FP)
Örnek: 100 kişiye hasta dedik, bunların 90'ı gerçekten hastaydı
Precision = 90/100 = 0.90
- Yüksek precision: Az sayıda yanlış alarm (false positive)


Recall
- Gerçekte pozitif olanların ne kadarını yakalayabildik?
- Formül: TP / (TP + FN)
Örnek: Gerçekte hasta olan 100 kişinin 80'ini yakalayabildik
Recall = 80/100 = 0.80
- Yüksek recall: Az sayıda kaçırdığımız hasta (false negative)

F1 Score
- Precision ve Recall'un harmonik ortalaması
- Formül: 2 × (Precision × Recall) / (Precision + Recall)
- Her iki metriği de dengeli bir şekilde değerlendirir


Gerçek hayattan bir örnek:

    Spam e-posta filtreleme:
        - Precision: Spam dediğimiz e-postaların gerçekten spam olma oranı
        - Recall: Tüm spam e-postaların ne kadarını yakalayabildiğimiz
        - F1 Score: İkisinin de dengeli bir değerlendirilmesi

# PRC (Precision-Recall Curve)

- Precision'a karşı Recall'u gösterir.
- AUPRC: Eğrinin altında kalan alan
- Dengesiz veri setlerinde daha iyi bir değerlendirme metriği

# ROC (Receiver Operation Characteristic) Curve

- True Positive Rate'e karşı False Positive Rate'i gösterir
- TPR = TP/(TP+FN) # Recall/Sensitivity
- FPR = FP/(FP+TN) # 1 - Specifity
- AUC-ROC: Eğrinin altında kalan alan (Area Under Curve)
- Dengeli veri setlerinde daha iyi çalışır

# Temel Farklar:

- ROC: Sınıflar dengeli olduğunda kullanışlı
- PRC: Dengesi veri setlerinde daha iyi
- PRC. false positive'lere karşı daha hassas
- ROC, false negative'lere karşı daha hassas
