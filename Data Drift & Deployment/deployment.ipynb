Unit Test Nedir?

Unit test, yazılımın en küçük test edilebilir parçalarının (genellikle tek bir fonksiyon veya metodun) izole bir şekilde test edilmesidir. Bir fonksiyona belirli girdiler verip, beklenen çıktıları alıp almadığımızı kontrol ederiz.

Örneğin, bir toplama fonksiyonu yazdıysak, bu fonksiyona 2 ve 3 verdiğimizde 5 döndürüp döndürmediğini test ederiz. Bu testler otomatiktir ve kodda değişiklik yaptığımızda her şeyin hala beklendiği gibi çalışıp çalışmadığını hızlıca kontrol etmemizi sağlar.

Böylece kodumuzda bir değişiklik yaptığımızda, var olan fonksiyonları bozmadığımızdan emin oluruz ve hataları erken aşamada yakalarız.

# Neden buna ihtiyacımız var?

ML projemizde bir fonksiyonu yazdığımızı düşünelim. Bu fonksiyonun doğru çalıştığından emin olmamız gerekir. Bu aşamada unit test ten faydalanırız. Fonksiyona bildiğimiz girdileri verip, beklediğimiz sonuçları alıp almadığımızı kontrol ederiz

ML Projelerinde FastAPI kullanmak

FastAPI'yi şöyle düşünelim: Harika bir ML modelimiz var ve bunu bir web servisi olarak sunmak istiyoruz. Yani internetten gelen isteklere modelinizin cevap vermesini istiyorsunuz. 

Otomatik dokümantasyon oluşturması: kodumuzu yazdıktan sonra, FastAPI sizin için otomatik olarak bir API dokümantasyonu oluşturuyor. Bu dokümantasyonu hem insanlar okuyabiliyor, hem de başka sistemler kullanabiliyor.

Veri doğrulama: FastAPI veri doğrulamayı otomatik olarak yapıyor. Yanlış formatta veri gelirse, daha modele ulaşmadan hata veriyor.Type hints desteği sayesinde kodun daha okunabilir ve hata ayıklamasının daha kolay olmasını sağlıyor
Özetle FastAPI, ML modellerinizi web servisi olarak sunmanın en modern ve hızlı yollarından biri. Hem geliştirmesi kolay, hem de performansı yüksek.


Alternatiflerine Göre Avantajları

Flask'e Göre
- Daha yüksek performans
- Modern özellikler
- Otomatik dokümantasyon
- Async yapı

Django'ya Göre
- Daha hafif
- ML odaklı
- Hızlı deployment
- Düşük overhead

ML Projelerinde Docker Kullanımı

Docker'ı şöyle düşünelim: Elinizde bir ML modeliniz var ve bunu farklı ortamlarda çalıştırmanız gerekiyor. Tıpkı bir kutuya tüm eşyalarınızı koyup taşıdığınız gibi, Docker da modelinizi ve ihtiyaç duyduğu her şeyi (Python versiyonu, kütüphaneler, ayarlar vs.) bir container içine paketliyor.

Ölçeklenebilirlik
- Modele gelen istekler arttığı zaman Docker sayesinde aynı container'dan birden fazla kopyayı hızlıca oluşturup, yükü dengeleyebiliriz.

Özetle Docker, ML modellerinizi "paketle ve çalıştır" mantığıyla her yerde aynı şekilde çalışmasını sağlayan bir teknoloji. Bu da biz geliştiricilerin hayatını önemli ölçüde kolaylaştırıyor
